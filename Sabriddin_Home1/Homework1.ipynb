```python
import pandas as pd
from scipy.stats import shapiro, levene, ttest_ind
```

```python
titanic_df = pd.read_csv('Titanic-Dataset.csv')
titanic_df = titanic_df.dropna(subset=['Age'])
```

```python
groups = {
    'men_survived': titanic_df.query("Sex == 'male' and Survived == 1"),
    'men_not_survived': titanic_df.query("Sex == 'male' and Survived == 0"),
    'women_survived': titanic_df.query("Sex == 'female' and Survived == 1"),
    'women_not_survived': titanic_df.query("Sex == 'female' and Survived == 0"),
}
```

```python
from numpy.random import default_rng
rng = default_rng()
for label, data in groups.items():
    sample_means = [data['Age'].sample(50, replace=True, random_state=rng.integers(0, 1e6)).mean() for _ in range(399)]
    result = shapiro(sample_means)
    print(f"{label}: Shapiro-Wilk test -> stat={result.statistic:.5f}, p-value={result.pvalue:.5f}")
```

```python
samples = {key: val['Age'].sample(50, random_state=rng.integers(0, 1e6)) for key, val in groups.items()}

lev_stat1 = levene(samples['men_survived'], samples['women_survived'])
print(f"Levene (survived): stat={lev_stat1.statistic:.5f}, p-value={lev_stat1.pvalue:.5f}")
```

```python
lev_stat2 = levene(samples['men_not_survived'], samples['women_not_survived'])
print(f"Levene (not survived): stat={lev_stat2.statistic:.5f}, p-value={lev_stat2.pvalue:.5f}")
```

```python
t_stat1 = ttest_ind(samples['men_survived'], samples['women_survived'], equal_var=True)
print(f"T-test (survived): stat={t_stat1.statistic:.5f}, p-value={t_stat1.pvalue:.5f}")
```

```python
t_stat2 = ttest_ind(samples['men_not_survived'], samples['women_not_survived'], equal_var=True)
print(f"T-test (not survived): stat={t_stat2.statistic:.5f}, p-value={t_stat2.pvalue:.5f}")
```
